6-Importation des bibliothèques pour analyser les sentiments dans la colonne text de notre dataset

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
def analyse(texte):
    sia=SentimentIntensityAnalyzer() #crée un objet ... pour analyser le sentiment du texte
    sentiment_scores=sia.polarity_scores(texte) #Analyse le texte et retourne un dictionnaire avec les scores des sentiments
    if sentiment_scores["compound"]>=0.05:      #Si le score composite de sentiment est supérieure ou égale à 0,05 ,la fonction retourne positif
        return "positif"
    elif sentiment_scores["compound"]<=-0.05:
         return "negatif"
    else :
        return "neutre"
```

Cette fonction ⁠"analyse "⁠ prend un texte en entrée et retourne une étiquette de sentiment ("positif", "negatif" ou "neutre") en fonction de l'analyse du sentiment effectuée par l'outil VADER. le score composite de sentiment est une valeur qui répresente l'intensité du sentiment exprimé dans le texte .

```python
def plot_sentiment_proportions(datas):
    sentiment_counts = data['sentiment'].value_counts()
    plt.figure(figsize=(8, 6))
    sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values)
    plt.title('Proportion of Sentiments in Posts')
    plt.xlabel('Sentiment')
    plt.ylabel('Number of Posts')
    plt.show()

plot_sentiment_proportions(data)
```

```python

def plot_sentiment_evolution_years(data):

    sentiment_by_year = data.groupby('Year')['intensité'].mean()
    plt.figure(figsize=(10, 6))
    plt.plot(sentiment_by_year.index, sentiment_by_year.values)
    plt.xlabel('Year')
    plt.ylabel('Mean Sentiment Intensity')
    plt.title('Evolution of Sentiment Intensity Over Time')
    plt.grid(True)
    plt.show()

plot_sentiment_evolution_years(data)
```

```python
def  plot_sentiment_evolution_months(data):
    sentiment_by_month = data.groupby('Month')['intensité'].mean()

    plt.figure(figsize=(10, 6))
    plt.plot(sentiment_by_month.index, sentiment_by_month.values)
    plt.title('Sentiment Intensity Over Time (Monthly)')
    plt.xlabel('Month')
    plt.ylabel('Average Sentiment Intensity')
    plt.grid(True)
    plt.show()
plot_sentiment_evolution_months(data)
```

```python
def sentiment_countries(data):
    if 'Country' in data.columns:
      sentiment_by_country = data.groupby('Country')['intensité'].mean()

      plt.figure(figsize=(15, 6))
      sns.barplot(x=sentiment_by_country.index, y=sentiment_by_country.values)
      plt.title('Average Sentiment Intensity by Country')
      plt.xlabel('Country')
      plt.ylabel('Average Sentiment Intensity')
      plt.xticks(rotation=90)
      plt.tight_layout()
      plt.show()
    else:
      print("No 'Country' column found in the DataFrame. Please ensure your CSV file includes a 'Country' column.")
    
sentiment_countries(data)
```

```python
#proportions des sentiments pour un post id donne 
import matplotlib.pyplot as plt

def sentiment_proportions(dataframe, post_id, global_sentiment):
    count_positif = 0
    count_negatif = 0
    count_neutre = 0

    # Filtrer les commentaires associés au post_id donné
    comments_df = dataframe[dataframe['post_id'] == post_id]

    # Parcourir les commentaires et incrémenter le bon compteur
    for sentiment in comments_df['sentiment']:
        if sentiment == 'positif':
            count_positif += 1
        elif sentiment == 'negatif':
            count_negatif += 1
        elif sentiment == 'neutre':
            count_neutre += 1

    # Stocker les résultats dans un dictionnaire
    sentiment_counts = {
        'positif': count_positif,
        'negatif': count_negatif,
        'neutre': count_neutre
    }

    # Affichage des résultats
    print(f"Pour le post_id {post_id} avec sentiment global '{global_sentiment}':")
    print(f"Nombre de commentaires positifs : {count_positif}")
    print(f"Nombre de commentaires négatifs : {count_negatif}")
    print(f"Nombre de commentaires neutres : {count_neutre}")

    # Affichage du nombre total de commentaires
    total_comments = count_positif + count_negatif + count_neutre
    print(f"Nombre total de commentaires : {total_comments}")

    # Calcul des proportions en pourcentage
    proportion_positif = (count_positif / total_comments) * 100
    proportion_negatif = (count_negatif / total_comments) * 100
    proportion_neutre = (count_neutre / total_comments) * 100

    # Affichage des proportions sous forme d'histogramme
    type_sentiment = ['positif', 'negatif', 'neutre']
    proportions = [proportion_positif, proportion_negatif, proportion_neutre]
    colors = ['#FF0000', '#0000FF', '#FFFF00']

    # Création de l'histogramme
    plt.bar(type_sentiment, proportions, color=colors)
    plt.xlabel("Type de Sentiment")
    plt.ylabel("Proportion (%)")
    plt.title(f"Proportion des sentiments pour le post_id {post_id} (sentiment global : {global_sentiment})")
    plt.show()
```

```python
#ressort les proportions des sentiments pour un tag precis
def sentiment_proportions_par_tag(dataframe,tag):
    # Filtrer les posts par tag
    posts_tag = dataframe[dataframe['tag'].str.lower() == tag.lower()]


    # Filtrer les commentaires pour les posts associés à ce tag
    comments_tag = dataframe[dataframe['post_id'].isin(posts_tag['post_id'])]


    # Calcul des proportions des sentiments
    count_positif = comments_tag[comments_tag['sentiment'] == 'positif'].shape[0]
    count_negatif = comments_tag[comments_tag['sentiment'] == 'negatif'].shape[0]
    count_neutre = comments_tag[comments_tag['sentiment'] == 'neutre'].shape[0]

    total_comments = count_positif + count_negatif + count_neutre

    proportion_positif = (count_positif / total_comments) * 100
    proportion_negatif = (count_negatif / total_comments) * 100
    proportion_neutre = (count_neutre / total_comments) * 100

    # Affichage des résultats
    print(f"\nProportions des sentiments pour le tag '{tag}':")
    print(f"Nombre de commentaires positifs : {count_positif} ({proportion_positif:.2f}%)")
    print(f"Nombre de commentaires négatifs : {count_negatif} ({proportion_negatif:.2f}%)")
    print(f"Nombre de commentaires neutres : {count_neutre} ({proportion_neutre:.2f}%)")

    # Affichage de l'histogramme des proportions
    sentiments = ['positif', 'negatif', 'neutre']
    proportions = [proportion_positif, proportion_negatif, proportion_neutre]
    colors = ['#FF0000', '#0000FF', '#FFFF00']  # Couleurs pour les sentiments

    plt.bar(sentiments, proportions, color=colors)
    plt.xlabel('Sentiment')
    plt.ylabel('Proportion (%)')
    plt.title(f'Proportions des sentiments pour le tag "{tag}"')
    plt.show()

```
